```python
def meters_to_cm(meters_value):
    """
    Convert a value from meters to centimeters.
    Args: 
        meters_value (float): The distance or length in meters.
    Returns:
        centimeters_result(float): The equivalent distance in centimeters.
    """

    centimeters_result = meters_value * 100
    return centimeters_result
```

```python
help(meters_to_cm)
```

```python
print(meters_to_cm(2.5)) # retorna 250.0)
```

```python
def create_user(name: str, age: int, city: str) -> dict:
    """
    This function creates a user profile with the provided name, age and city by the user.
    Arguments:
    name(str): The name of the user.
    age(int): The age of the user.
    city(str): The city where the user lives.
    returns:
    user_data(dict): A dictionary containing the user's name, age and city.
    """
    user_data = {
        'name': name,
        'age': age, 
        'city': city
    }

    if age <= 0:
        return 'Age must be a positive interger.'
    
    return user_data
```

```python
user = create_user(name='Vinicius', age=27, city='Porto Alegre')
print(user)
```

```python
user = create_user(name='Alice', age=-26, city='Gramado')
print(user)
```

### Try & Except
 - except NomeDoErro: Trata um erro em específico
 - except Exception: Tratando qualquer tipo de erro de maneira genêrica
 - as e: Guardar na var "e" a mensagem de erro que o Python gerou

```python
# Exemplo de um except específico simples
def realizar_divisao(numero, divisor):
    try:
        resultado = numero / divisor
        print(f'O resultado da divisão é: {resultado}')

    except ZeroDivisionError:
        print('Erro: Não foi possível dividir por 0')

# --- Testes ----
```

```python
realizar_divisao(10, 2) # Retorna 5.0
```

```python
realizar_divisao(10, 0) # Retorna 'Erro: Não foi possível dividir por 0, lembrando que 
                            # em Python não existe divisão por zero.
```

```python
# Exemplo com mais de except específico
def acessar_lista(indice: int) -> None:
    cores = ['Verde', 'Amarelo', 'Azul', 'Vermelho']

    try:
        print(f"A cor escolhida foi: {cores[indice]}")
    except IndexError:
        print(f"Erro: O índice {indice} está fora da lista")
    except TypeError:
        print("Erro: Você precisa digitar um número inteiro para acessar a lista")
```

```python
acessar_lista(1) # Retorna 'A cor escolhida foi: amarelo'
```

```python
acessar_lista(10) # Retorna 'Erro o índice 10 está fora da lista'
```

```python
def acessar_lista_generica(indice: int) -> None:
    cores = ['Verde', 'Amarelo', 'Azul', 'Vermelho']

    try:
        print(f"A cor escolhida foi: {cores[indice]}")
    except Exception as e:
        print(f"Ocorreu um erro ao tentar acessar a lista!")
        print(f"Detalhe técnico do erro: {e}")
```

```python
def dividir_conta_total(total: str, pessoas: str) -> float | str:
    try:
        valor_total = float(total)
        quantidade_pessoas = int(pessoas)

        return valor_total / quantidade_pessoas

    except ValueError: 
        return 'Erro: Os valores de entrada devem ser números.'
    
    except ZeroDivisionError:
        return 'Erro: A quantidade de pessoas deve ser maior que zero para poder dividir a conta.'
    
    except Exception as e:
        return f"Erro inesperado: {e}"
    
resultado1 = dividir_conta_total("200", "4")
print(resultado1) 

resultado2 = dividir_conta_total("200", "0")
print(resultado2)

resultado3 = dividir_conta_total(None, "5")
print(f"Resultado 3: {resultado3}")
```

```python
print(dividir_conta_total(total=30, pessoas=10)) # Retorna 3.0
```

```python
print(dividir_conta_total(total=10, pessoas=0)) # Retorna 'Erro: Não é possível dividir por zero.'
```

```python
print(dividir_conta_total(None, None)) # Retorna 'Erro: Por favor insira um valor númerico válido para o total da conta.'
```

## Raise

```python
def realizar_saque(valor):
    saldo = 100

    try:
        if valor > saldo:
            raise ValueError('Saldo insuficiente para realizar o saque.')
        
        print(f'Saque de R${valor} realizado com sucesso. \nSaldo restante: R${saldo - valor}')
    
    except Exception as e:
        print(f'Erro: {e}')
```

```python
realizar_saque(200)
```

```python
realizar_saque(100)
```
