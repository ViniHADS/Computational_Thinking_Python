# Funções PT.1

```python
## bloco de instruções que executa uma tarefa específica.

# Vantagens:
# - Reutilização de código: Funções permitem que você escreva um bloco de código uma vez e o reutilize em diferentes partes do programa, evitando a duplicação de código.
# - Organização: Funções ajudam a organizar o código, tornando-o mais legível e fácil de entender.
# - Modularidade: Funções permitem que você divida um programa em partes menores e mais gerenciáveis, facilitando a manutenção e a depuração.
# - Manuntenção: Se você precisar fazer alterações em um bloco de código, pode fazê-lo em um único lugar (a função), em vez de procurar e modificar o código em vários lugares do programa.
```

```python
# Sintaxe básica de uma função em Python:
def somar(a, b):
    resultado = a + b
    print(f"A soma de {a} e {b} é: {resultado}")

# Exemplo de uso da função:
somar(5, 3)  # Saída: A soma de 5 e 3 é: 8
```

```python
# Variaveis LOCAIS: São aquelas que são definidas dentro de uma função e só podem ser acessadas dentro dessa função. 
# Elas são criadas quando a função é chamada e destruídas quando a função termina sua execução.
# Variaveis GLOBAIS: São aquelas que são definidas fora de qualquer função e podem ser acessadas por qualquer parte do programa, 
# incluindo dentro de funções. Elas permanecem na memória durante toda a execução do programa.  
```

```python
# Exemplo de código com variáveis LOCAIS:

def calcular_area_circulo(raio):
    pi = 3.14159  # Variável local
    area = pi * (raio ** 2)
    print(f"A área do círculo com raio {raio} é: {area}")
calcular_area_circulo(5)  # Saída: A área do círculo com raio 5 é: 78.53975 
```

```python
# Exemplo de variáveis GLOBAIS:

def exibir_nome():
    print(f"Meu nome é {nome}")  # Acessando a variável global 'nome'
nome = "Alice"  # Variável global
```

```python
exibir_nome()  # Saída: Meu nome é Alice

# -------- Fim do exemplo de código com variáveis GLOBAIS ---------
```

```python
# Função que analisa tamanho de senhas

def analisar_senhas(lista_de_senhas):
    for senha in lista_de_senhas:
        # Verificar se o comprimento da str é menor que 6
        if len(senha) < 6:
            print(f'{senha}: Muito curta (risco de segurança)')
        else:
            print(f'{senha} Tamanho Ok')
```

```python
senhas_usuarios = ["240599", "vini", "Avantasia99"]

analisar_senhas(senhas_usuarios)
```

```python
# Scanner de desconto

produto1 = {"nome": "teclado", "preco": 100.0, "categoria" : "Eletrônicos"}

def verificar_produto(item):
    nome = item['nome']
    preco = item['preco']
    cat = item['categoria']

    print(f'Analisando {nome}...')

    if cat == "Eletrônicos":
        desconto = preco * 0.10
        final = preco - desconto
        print(f'Promoção: Novo preço de {nome} é R${final}')
    else: 
        print(f'Não tem desconto. Preço é {preco}')

verificar_produto(produto1)
```

### Desafio: Frete inteligente

```python
# ## Desafio: Frete Inteligente

#Contexto: Sua loja virtual oferece frete grátis para compras acima de um certo valor. Para compras menores, é cobrada uma taxa fixa. 
# Você precisa criar uma função que receba o valor da compra e decida qual o valor final que o cliente deve pagar.
#O que deve ser feito:
# - Crie uma função chamada calculate_shipping.
# - Essa função deve receber um parâmetro: purchase_amount.

# Dentro da função:
# - Se o valor da compra for maior ou igual a 200 reais, o frete é R$ 0,00.
# - Se for menor que 200 reais, o frete é uma taxa fixa de R$ 20,00.
# - A função deve retornar (return) apenas o valor do frete.

# No programa principal:
# - Peça ao usuário o valor total do carrinho de compras (via input).
# - Chame a função para descobrir o valor do frete.
# - Calcule o total final (valor da compra + frete).
# - Exiba os detalhes para o usuário: print(f"\nResumo do Pedido:") print(f"Compra: {}") print(f"Frete: {}") print(f"Total: {}")
```

```python
# Criando a função que calcula o frete:
def calculo_frete(purchase_amount):
    # Verifica se a compra é maior ou igual a 200
    if purchase_amount >= 200:
        frete = 0 # Se sim, frete é grátis
    else:
        frete = 20 # Se não, frete custa R$20

    # Devolve o valor do frete para quem chamou a função
    return frete

# Pede para o usuário digitar o valor da compra e converte para número decimal
purchase_amount = float(input("Digite o valor da compra:    "))

# Chama a função e guarda o resultado na variável frete
frete = calculo_frete(purchase_amount)

# Soma o valor da compra com o frete para obter o total
total = purchase_amount + frete

# Exibe o resumo do pedido
print(f'\nResumo do pedido:')
print(f'Compra: R${purchase_amount}')
print(f'Frete: R${frete}')
print(f'Total: R${total}')
```

### Desafio prático: Currency Converter

```python
#Você está desenvolvendo o módulo financeiro de um sistema internacional. Todos os valores no seu banco de dados devem ser armazenados em Dólar (USD). 
# O usuário pode inserir valores em Real (BRL) ou Euro (EUR). Sua missão é criar uma função que trate o texto digitado pelo usuário, converta para o câmbio correto 
# e retorne o valor pronto para ser salvo.
# O que a função (convert_to_usd) deve fazer:
# Parâmetros: amount (valor) e currency_type (a moeda de origem).
# Substituir vírgulas por pontos (para que o Python entenda como número decimal).
# Remover qualquer caractere que não seja número ou ponto.
# Converter o texto limpo para o tipo decimal (float).

def convert_to_usd(amount, currency_type):
    clean_amount = amount.replace(',', '.')
    value = float(clean_amount)

    if currency_type.upper() == "BRL":
        usd_value = value / 5.00 # taxa de 1U$D = 5 Reais
    elif currency_type.upper() == "EUR":
        usd_value = value * 1.10 # Taxa 1 EUR = 1.1 U$D
    else:
        usd_value = value
    
    return usd_value

print("-" * 10)
user_value = input('Coloque o valor a ser convertido:   ')
user_currency = input('Moeda origem (BRL/EUR):  ')
print("-" * 10)


final_price_usd = convert_to_usd(user_value, user_currency)

print(f'\n valor ')
```
