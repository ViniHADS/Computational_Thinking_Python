```python
import re

def convert_to_usd(amount, currency_type):
    clean_amount = re.sub(r'[^0-9,.]', '', amount) # Sintaxe da lib rejex, aqui no caso está substituindo todo carectere que não for número a número e só aceita números de 0 à 9.

    clean_amount = clean_amount.replace(',', '.')

    value = float(clean_amount)

    currency = currency_type.upper().strip()

    if currency == 'BRL':
        usd_value = value / 5.0

    elif currency == 'EUR':
        usd_value = value * 1.10

    else:
        usd_value = value

    return usd_value

print('- - - - CURRENCY CONVERTER TOOL - - - -')
user_value = input('Coloque o valor a ser convertido:    ')
user_currency = input('Moeda (BRL/EUR):     ')

final_price_usd = convert_to_usd(user_value, user_currency)
print(f'\nValor final a ser salvo no banco de dados: {final_price_usd}')
```
