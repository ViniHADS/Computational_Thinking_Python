## JSON

```python
# Exemplo código Python de dict vs JSON:
# Dict Python:
usuario = {
    "nome": "Vinicius",
    "idade": 27,
    "estudante": True,
    "linguagens": ["Python", "C#", "JavaScript"]
}

print(usuario["nome"])
```

```python
# Verificação Python:
print(type(usuario))
```

```python
# JSON mesmo exemplo
{
    "nome": "Vinicius",
    "idade": 27,
    "estudante": true,
    "linguagens": ["Python", "C#", "JavaScript"]
}
```

```python
import json

# Python dict
usuario = {
    "nome": "Vinicius",
    "idade": 27,
    "estudante": True,
    "linguagens": ["Python", "C#", "JavaScript"]
}

# Converte o dict Python para uma string no formato JSON
json_usuario = json.dumps(usuario)

print(json_usuario)
print(type(json_usuario))
```

```python
import json

# 1. Criando um dict do Python
usuario = {
    "nome": "Vinicius",
    "idade": 27,
    "estudante": True,
    "linguagens": ["Python", "C#", "JavaScript"]
}

# 2. Criando um arquivo JSON e salvando o dict nele
with open("usuario.json", "w", encoding="utf-8") as arquivo:
    json.dump(usuario, arquivo, ensure_ascii=False, indent=4)

print("Arquivo JSON criado com sucesso!")
```

# JSON -> Python

```python
texto = '''
{
    "nome" : "Jean",
    "idade" : 32,
    "curso" : "Python"
}
'''
```

```python
aluno = json.loads(texto)

print(aluno)
print(type(aluno))
```

```python
curso = [
    "Python",
    "SQL",
    "JavaScript"
]
```

```python
# List python para JSON do tipo array

json_curso = json.dumps(curso, indent=4)

print(json_curso)
```

```python
# Lista de objeto -> Array de objetos

alunos = [
    {
        "nome" : "Luis",
        "idade" : 22
    },
    {
        "nome" : "Vitória",
        "idade" : 18
    },
    {
        "nome" : "João",
        "idade" : 18
    }
]
```

```python
#Transformando em json

json_alunos = json.dumps(alunos, indent=4, ensure_ascii=False)

print(json_alunos)
```

```python
# Criando arquivos JSON

with open('alunos.json', 'w', encoding='utf-8') as arquivo:
    json.dump(alunos, arquivo, indent=4, ensure_ascii=False)
```

```python
# Desserialização
with open('alunos.json', 'r', encoding='utf-8') as arquivo:
    dados = json.load(arquivo)
```

```python
print(dados)
```

```python
type(dados)
```

```python
# E aqui você já consegue manipular isso aqui igual uma lista normal de python.
print(dados[1])
```

```python
textos1 = '''
{
    "nome" : "Maria",
    "idade" : 30
}
'''
```

```python
try: 
    professor = json.loads(textos1)
    
except Exception:
    print('JSON Inválido')
```

```python
try:
    professor = json.loads(textos1)

except json.JSONDecodeError as e:
    print(f'JSON Inválido. Erro: {e}')
```

```python
import json

# Criando um JSON enunciado

#Crie um dict chamado produto com:
produto = {
    "id": 1,
    "nome": "Teclado Mecânico",
    "preco": 299.90,
    "estoque": 15
}

# Convertendo o dict para JSON com indentação de 4 espaços
produto_json = json.dumps(produto, indent=4, ensure_ascii=False)

# Mostrando o resultado
print(produto_json)
```

```python
import json

texto = '''{ "nome": "Ana", "idade": 21, "curso": "Python" }'''

# Converte o JSON (string) para um dict Python
dados = json.loads(texto)
```

```python
# Mostra o nome e o curso
print(dados["nome"])
print(dados["curso"])
type(dados)
```
