## Projeto Prático - Data App

```python
# JSON Pt.2
import pandas as pd
import json
```

```python
dados_dict = {
    "nome" : ["Ana", "Bruno", "Carla"],
    "idade" : [25, 30, 22],
    "cidade" : ["São Paulo", "Rio de Janeiro", "Curitiba"]
}
```

```python
with open ('json_data.json', 'w', encoding="utf-8") as file:
    json.dump(dados_dict, file, ensure_ascii=False, indent=4)
```

```python
with open('json_data.json', 'r', encoding="utf-8") as file:
    dados_python = json.load(file)
```

```python
print(dados_python)
type(dados_python)
```

```python
# Abordagem A: pd.dataframe
df_opcao_a = pd.DataFrame(dados_python)
df_opcao_a
```

```python
# Abordagem B: read_json
df_opcao_b = pd.read_json('json_data.json')
df_opcao_b
```

```python
df_opcao_b.to_csv('usuario.csv', index=False, encoding="utf-8")
```

### CSV to JSON

```python
dados = {
    "id" : [1, 2, 3],
    "nome" : ["Bruno", "Silvio", "Carla"],
    "cargo" : ["Analista", "Desenvolvedor", "Gerente"],
    "salario" : [4500, 6200, 9500]
}

df_original = pd.DataFrame(dados)
df_original
```

```python
df_original.to_csv("dados_exemplo.csv", index=False, encoding="utf-8")
```

```python
df_original.to_json(
    "dados_output.json", 
    orient="records",
    indent=4,
    force_ascii=False
)
```

```python
df_original.to_json(
    "dados_output2.json", 
    orient="columns",
    indent=4,
    force_ascii=False
)
```

```python
df_original.to_json(
    "dados_output3.json", 
    orient="table",
    indent=4,
    force_ascii=False
)
```
