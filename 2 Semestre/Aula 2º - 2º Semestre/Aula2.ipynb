# Comando With

```python
with open('resultado.txt', 'w', encoding='utf-8') as arquivo:
    arquivo.write('primeira linha \n Segunda linha \n')
```

```python
with open('resultado.txt', 'w', encoding='utf-8') as arquivo:
    arquivo.write('Novo processamento')
```

```python
alunos = [
    "Pedro"
    "Vinicius"
    "Lucas"
    "Mario"
]

with open('alunos.txt', 'w', encoding='utf-8') as arquivo:
    for aluno in alunos:
        arquivo.write(aluno + '\n')
```

```python
alunos = [
    "George"
    "Ryan"
]

with open('alunos.txt', 'w', encoding='utf-8') as arquivo:
    for aluno in alunos:
        arquivo.write(aluno + '\n' )
```

```python
# Transformação de dados
with open ('alunos.txt', 'r', encoding='utf-8') as arquivo:
    for linha in arquivo:
        nome = linha.strip().upper()
        print(nome)
```

```python
with open ('alunos.txt', 'r', encoding='utf-8') as entrada:
    with open('alunos_uppercase.txt', 'w', encoding='utf-8') as saida:
        for linha in entrada:
            nome = linha.strip().upper()
            saida.write(nome + '\n')
```
