### Baú de Itens

```python
#Imagine que você está analizando dados de heróis de um jogo de RPG e cada um carrega
#uma lista de itens na mochila.
#Enunciado simples:
#Você tem um dicionário onde a chave é o nome do herói e o valor é a lista de itens.
#mochilas = {'Aragorn': ['Espada','Escudo','Poção'], 'Legolas': ['Arco','Flechas'],
#'Gimli':['Machado','Poção']}
#Seu objetivo é percorrer o grupo e avisar quais heróis estão carregando uma 'Poção'.
#Modelo: print(f'O herói{heroi} possui uma poção.)

mochilas = {
    'Aragorn': ['Espada', 'Escudo', 'Poção'],
    'Legolas': ['Arco', 'Flechas'],
    'Gimli': ['Machado', 'Poção']
}   
for heroi, itens in mochilas.items():
    if 'Poção' in itens:
        print(f'O herói {heroi} possui uma poção.')
    elif 'Poção' not in itens:
        print(f'O herói {heroi} não possui uma poção.')
```

### Reposição de estoque de Livros:

```python
#Enunciado: Você possui um dicionário chamado biblioteca onde cada chave é um 
# ISBN e o valor é outro dicionário com os detalhes do livro. 
# Sua missão é criar um script que percorra esses dados e gere apenas uma lista simples 
# (chamada livros_para_comprar) contendo os títulos dos livros que estão 
# indisponíveis atualmente no acervo para fazer um pedido de compras.

#biblioteca = { "978-01": {"titulo": "Python Fluente", "autor": "Luciano Ramalho", 
# "disponivel": True}, "978-02": {"titulo": "Código Limpo", "autor": "Robert Martin", 
# "disponivel": False}, "978-03": {"titulo": "Design Patterns", "autor": "GoF",
#  "disponivel": True}, "978-04": {"titulo": "Algoritmos", "autor": "Cormen", 
# "disponivel": False} }

biblioteca = {
    "978-01": {"titulo": "Python Fluente", "autor": "Luciano Ramalho", "disponivel": True},
    "978-02": {"titulo": "Código Limpo", "autor": "Robert Martin", "disponivel": False},
    "978-03": {"titulo": "Design Patterns", "autor": "GoF", "disponivel": True},
    "978-04": {"titulo": "Algoritmos", "autor": "Cormen", "disponivel": False}
}   
livros_para_comprar = []
for isbn, detalhes in biblioteca.items():
    if not detalhes['disponivel']:
        livros_para_comprar.append(detalhes['titulo'])
print("Livros indisponíveis para compra:")
for livro in livros_para_comprar:
    print(f"- {livro}") 

for isbn, detalhes in biblioteca.items():
    if detalhes['disponivel']:
        livros_para_comprar.append(detalhes['titulo'])
print("Livros disponíveis para compra:")
for livro in livros_para_comprar:
    print(f"- {livro}")
```
